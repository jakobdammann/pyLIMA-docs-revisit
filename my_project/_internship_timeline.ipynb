```python
import numpy as np
import matplotlib.pyplot as plt

schedule = np.genfromtxt('_schedule.csv', delimiter=',')
weeks = schedule[:,0]
hours = schedule[:,2]

unique_weeks, idx = np.unique(weeks, return_inverse=True)
hours_per_week = np.bincount(idx, weights=hours)

plt.figure(figsize=(8,5))
plt.title(f'Total hours: {np.sum(hours)}')
plt.bar(unique_weeks, hours_per_week)
plt.ylabel('hours')
plt.xlabel('week')
plt.show()
```
