```python
import numpy as np
import matplotlib.pyplot as plt

schedule = np.genfromtxt('_schedule.csv', delimiter=',')
weeks = schedule[:,0]
hours = schedule[:,2]

plt.figure(figsize=(8,5))
plt.title(f'Total hours: {np.sum(hours)}')
plt.bar(weeks, hours)
plt.ylabel('hours')
plt.xlabel('week')
plt.show()
```
